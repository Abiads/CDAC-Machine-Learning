### Entropy
What is entropy?
Measure of Impurity, disorder or uncertainty in a dataset.

In the context of ML, entropy is used in decision trees to determine the best feature to split the data.

The feature that results in the highest information gain (reduction in entropy) is chosen for the split.

Example :
- Dataset with 10 samples of same constants (apples) : Entropy = 0
- Dataset with 5 apples and 5 orangesc (50/50) : Entropy = 1
- Dataset with 7 apples and 3 oranges : Entropy = 0.881
- Dataset with 9 apples and 1 orange : Entropy = 0.468

Math Formula for Entropy:
$H(X) = -\sum P(x) \log_2 P(x)$


```python
# entropy using pure python
import math
from collections import Counter
```

```python
'''
formula for entropy is : 
H(X) = -∑ P(x) log2 P(x)
'''
def calculate_entropy(labesl):
    total_count = len(labesl)
    counts = Counter(labesl)
    entropy = 0.0
    for count in counts.values():
        probability = count / total_count
        entropy -= probability * math.log2(probability)
    return entropy
```

```python
pure_group = ['Iguana', 'Iguana', 'Iguana', 'Iguana']
mixed_group = ['Iguana', 'Iguana', 'Dog', 'Dog']
print("Entropy of mixed group:", calculate_entropy(mixed_group))
print("Entropy of pure group:", calculate_entropy(pure_group))
```

```python
# python library for entropy calculation
from scipy.stats import entropy
from collections import Counter

h_mixed = entropy([0.5,0.5],base=2)
print("Entropy of mixed group:", h_mixed) # should be 1.0 for a 50/50 distribution
h_pure = entropy([1.0,0.0],base=2)
print("Entropy of pure group:", h_pure) # should be 0.0 for a pure distribution
```

```python
# --- Case 1: Pure Group ---
pure_group = ['Iguana', 'Iguana', 'Iguana', 'Iguana']
pure_counts = list(Counter(pure_group).values())  
print("Pure Group Entropy:", entropy(pure_counts, base=2))

# --- Case 2: High Entropy (50/50 split) ---
mixed_group = ['Iguana', 'Iguana', 'Dog', 'Dog']
mixed_counts = list(Counter(mixed_group).values())  # [2, 2]
print("50/50 Mixed Group Entropy:", entropy(mixed_counts, base=2))

# --- Case 3: Low Entropy (9:1 split) ---
mixed_group_low = ['Iguana'] * 9 + ['Dog']
mixed_counts_low = list(Counter(mixed_group_low).values())  # [9, 1]
print("9:1 Mixed Group Entropy:", entropy(mixed_counts_low, base=2))
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,export_text
# Simple dataset: Wickets > 100 -> Bowler, else Batsman
df = pd.DataFrame([
    {"wickets": 5,   "role": "Batsman"},
    {"wickets": 8,   "role": "Batsman"},
    {"wickets": 337, "role": "Bowler"},
    {"wickets": 502, "role": "Bowler"}
])
X = df[["wickets"]]
y = df["role"]

clf = DecisionTreeClassifier(criterion="entropy", random_state=42)
clf.fit(X, y)

print("Predictions:", clf.predict(pd.DataFrame({"wickets": [10, 400]})))
print("_" * 60)
print("Decision Tree Structure:\n", export_text(clf, feature_names=["wickets"]))
```
