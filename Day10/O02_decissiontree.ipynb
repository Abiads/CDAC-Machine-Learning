```python
# Visualization setup
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11

print("Environment setup complete.")
```

```python
import numpy as np
counts = np.array([30, 70])
p = counts / np.sum(counts) # p = probabilities of each class
print("Class probabilities:", p)

# manual_gini
manual_gini = 1.0 - np.sum(p ** 2)
# manual_entropy
manual_entropy = -np.sum([pi * np.log2(pi) for pi in p if pi > 0])

impurity_df = pd.DataFrame({
    'Metric': ['Gini', 'Entropy'],
    "Formula": ["1 - Σp²", "-Σp log₂p"],
    "Theaoritical Minimum": [0.0, 0.0],
    "Theoretical Maximum": [0.5, 1.0],
    'Computed Value ([30,70]) split': [manual_gini, manual_entropy]
})
print(impurity_df.to_string(index=False))

# Plot Gini vs Entropy curves for binary probabilities p in [0, 1]
p_vals = np.linspace(0.001, 0.999, 500)
gini_curve = 1.0 - (p_vals**2 + (1 - p_vals)**2)
entropy_curve = -(p_vals * np.log2(p_vals) + (1 - p_vals) * np.log2(1 - p_vals))

plt.figure(figsize=(8, 4.5))
plt.plot(p_vals, gini_curve, label='Gini Impurity', lw=2.5)
plt.plot(p_vals, entropy_curve, label='Entropy (Entropy / 2 scaled for visual comparison)', lw=2.5, linestyle='--')
plt.title('Gini Impurity vs. Entropy for Binary Class Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Probability of Positive Class ($p$)')
plt.ylabel('Impurity Score')
plt.legend()
plt.tight_layout()
plt.show() 
```
